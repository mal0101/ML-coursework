```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
from sklearn.tree import DecisionTreeClassifier
from sklearn.neighbors import KNeighborsClassifier
```

```python
data = pd.read_csv("../../TP1/data/train_chiffres.csv").to_numpy()
```

```python
clf = DecisionTreeClassifier()
x=data[0:2100,1:]
label = data[0:2100,0]
clf.fit(x,label)
```

```python
xtest=data[2100:,1:]
labeltest=data[2100:,0]
p=clf.predict(xtest)
```

```python
count = 0
for i in range(len(p)):
    count += 1 if p[i] == labeltest[i] else 0

print("accuracy: ", count/len(labeltest))
```

```python
d=xtest[5]
nb_pixels_err = 100
for i in range(nb_pixels_err):
    position = np.random.randint(0, 784,1)[0]
    bruit = np.random.randint(-200, 200,1)[0]
    d[position] += bruit
    d[position] = d[position]%255
p=clf.predict([d])
print(p)
```

```python
d.shape=(28,28)
plt.imshow(255-d, cmap='gray')
plt.show()
```
