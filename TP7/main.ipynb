# TP 7: Les arbres de régression

Considérons le programme suivant qui utilise les arbres de décision pour un problème de régression.

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.tree import DecisionTreeRegressor
from sklearn.metrics import mean_squared_error
```

```python
np.random.seed(0)
X = np.sort(5 * np.random.rand(200, 1), axis=0)
y = np.sin(X)
plt.plot(X,y)
plt.plot(X,y)
plt.title("Courbe de répartition sinusoïdale")
```

```python
reg = DecisionTreeRegressor(max_depth=2)
reg.fit(X, y)
X_test = np.sort(5 * np.random.rand(80, 1), axis=0)
y_pred = reg.predict(X_test)
y_test = np.sin(X_test)
plt.figure()
plt.scatter(X,y, color="b", label="Exemples d'apprentissage")
plt.plot(X_test, y_pred, color="r", label='Prédiction', linewidth=3)
plt.xlabel("x")
plt.ylabel("y")
plt.title("Prédiction du modèle de régression")
plt.legend()
plt.show()
mse = mean_squared_error(y_test, y_pred)
print("Mean Squared Error:", mse)
```
