```python
import numpy as np
import pandas as pd
```

```python
data = pd.DataFrame([[0,1,'C1'],[0,0,'C1'],[1,1,'C2'],[1,0,'C2']], columns=['X1','X2','Class'])
data = data.to_numpy()
```

```python
data
```

```python
def entropy(S):
    distinct_classes = set(S[:,-1])
    classes = list(S[:,-1])
    s=0
    for c in distinct_classes:
        p = classes.count(c)/len(classes)
        s += -p*np.log2(p)
    return -1*s
entropy(data)
```

```python
def gain_info(S, n_col_att):
    classes = list(set(S[:,-1]))
    valeurs_attr = list(set(S[:,n_col_att]))
    Si = [[] for i in range(len(valeurs_attr))]
    for e in S:
        e = list(e)
        val_attr_pour_e = e[n_col_att]
        nb_sous_ensembles = valeurs_attr.index(val_attr_pour_e)
        Si[nb_sous_ensembles].append(e)
    Si = np.array(Si)
    som = 0
    for sous_ensemble in Si:
        som += (len(sous_ensemble)/len(S))*entropy(sous_ensemble)
    return abs(entropy(S) - som)
```

information gain is always positive

```python
print(gain_info(data, 0))
```

indice de gini

```python
def gini(S):
    clonne_des_classes = list(S[:,-1])
    classes = list(set(S[:,-1]))
    k = len(classes)
    s = 0
    for i in range(k):
        pi = clonne_des_classes.count(classes[i])/len(S[:,-1])
        s+= pi**2
    return 1-s
```

```python
data = pd.DataFrame([[0,1,'C1'],[0,0,'C1'],[1,1,'C2'],[1,0,'C2']], columns=['X1','X2','Class'])
data = data.to_numpy()
gini(data)
```

mesure de desordre

```python
def mesure_de_desordre(S, n_col_att):
    classes = list(set(S[:,-1]))
    valeurs_attribut=list(set(S[:,n_col_att]))
    Si = [[] for i in range(len(valeurs_attribut))]
    for e in S:
        valeurs_attribut_pour_e = e[n_col_att]
        nb_sous_ensembles = valeurs_attribut.index(valeurs_attribut_pour_e)
        Si[nb_sous_ensembles].append(e)
    Si = np.array(Si)
    som = 0
    for sous_ensemble in Si:
        som += (len(sous_ensemble)/len(S))*gini(sous_ensemble)
    return som
```
