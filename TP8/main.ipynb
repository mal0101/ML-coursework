```python
from collections import Counter
import math
def mean(labels):
    return sum(labels) / len(labels)
def mode(labels):
    return Counter(labels).most_common(1)[0][0]
def euclidean_distance(point1, point2):
    return math.sqrt(sum((a - b) ** 2 for a, b in zip(point1, point2)))
def knn(data, query, k, distance_utilise, choix_knn_regress_ou_classif):
    distances_et_positions = []
    for index, example in enumerate(data):
        distance = distance_utilise(example[:-1], query)
        distances_et_positions.append((distance, index))
    sorted_distances_et_positions = sorted(distances_et_positions)
    k_proches_distances_and_indices = sorted_distances_et_positions[:k]
    k_proches_classes = []
    for dist_indice in k_proches_distances_and_indices:
        num = dist_indice[1]
        k_proches_classes += [data[num][-1]]
    return k_proches_distances_and_indices, choix_knn_regress_ou_classif(k_proches_classes)
```

```python
from numpy import *
taille_poids = array([[167.66, 112.99], [182.38, 136.49], [176.97, 153.03],
                      [173.96, 142.24], [172.86, 136.49], [175.26, 149.03], [177.8, 154.94],
                      [171.45, 132.28], [178.39, 149.94], [175.26, 149.03], [177.8, 154.94],
                      [171.45, 132.28], [178.39, 149.94], [175.26, 149.03], [177.8, 154.94],
                      [171.45, 132.28], [178.39, 149.94], [175.26, 149.03]])

taille_en_cm = [161]
k = 3
distance_utilisee = euclidean_distance
choix_knn_regress_ou_classif = mean
R = knn(taille_poids, taille_en_cm, k, distance_utilisee, choix_knn_regress_ou_classif)
knn_distances_et_indices, knn_prediction = R
print("Knn implémenté pour regression:")
print(knn_distances_et_indices, knn_prediction)
from sklearn import neighbors
knn_regression = neighbors.KNeighborsRegressor(k)
X = [[e] for e in taille_poids[:,0]]
Y = [e for e in taille_poids[:,1]]
knn_regression.fit(X, Y)
print("Knn de sklearn pour regression:")
print(knn_regression.predict([taille_en_cm]))
```

```python
age_aimer_jeux_electroniques = array([[22,1], [23,1], [21,1], [18,1], [19,1],
                                      [25,0], [27,0], [29,0], [31,0], [45,0]])
age = [33]
distance_utilisee=euclidean_distance
choix_knn = mode
R = knn(age_aimer_jeux_electroniques, age, k, distance_utilisee, choix_knn)
knn_distances_et_indices, knn_prediction = R
print("Knn implémenté pour classification:")
print(knn_distances_et_indices, '\n', knn_prediction, '\n')
from sklearn.neighbors import KNeighborsClassifier
neigh = KNeighborsClassifier(n_neighbors=k, metric='euclidean')
X = [[e] for e in age_aimer_jeux_electroniques[:,0]]
Y = [e for e in age_aimer_jeux_electroniques[:,1]]
neigh.fit(X, Y)
print("Knn de sklearn pour classification:")
print(neigh.predict([age]))
```

```python
from sklearn.neighbors import KNeighborsClassifier
from sklearn.model_selection import GridSearchCV, train_test_split
import pandas as pd
import numpy as np
data = pd.read_csv('data/images_chiffres_codees_niveau_de_gris.csv')
data = np.array(data)
x=data[0:1000,1:]
y=data[0:1000,0]
X_train, X_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)
model = KNeighborsClassifier(n_neighbors=5, metric="manhattan")
model.fit(X_train, y_train)
score = model.score(X_test, y_test)
print("score sur test-set par k=5:", score)
distances = ['euclidean','manhattan']
valeurs_de_k = np.arange(1, 8)
parametres_grid = {"n_neighbors": valeurs_de_k, "metric": distances}
model = KNeighborsClassifier()
grid_search = GridSearchCV(model, parametres_grid, cv=3, scoring='accuracy')
grid_search.fit(X_train, y_train)
print("Meilleurs paramètres trouvés:", grid_search.best_params_)
model = grid_search.best_estimator_
model.fit(X_train, y_train)
score=model.score(X_test, y_test)
print("score sur test-set après GridSearchCV:", score)
```
